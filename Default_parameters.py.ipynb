```python
def power(base,exponent=2):
    return(base**exponent)
print(power(2))
print(power(2,3))
```

```python
def connect(host,port=3306,protocol='TCP'):
    return (f"{host} {port} {protocol}")
print(connect("amar"))
```

```python
def discount_price(price,discount=10):
    return price-(price*discount/100)
print(discount_price(10,20))
print(discount_price(10))
```

```python
#A default parameter is better than hardcoding because it allows the value to be changed when needed while still providing a default value.

def greet(name, message="Hello"):
    print(message, name)

greet("Amarnath")          # Hello Amarnath
greet("Amarnath", "Hi")    # Hi Amarnath
```
